```python
import requests
```

```python
def drugbank_to_chembl(drugbank_id):
    """Maps a DrugBank ID to a ChEMBL ID using the UniChem API."""
    # Source ID 2 = DrugBank, Source ID 1 = ChEMBL
    url = f"https://ebi.ac.uk{drugbank_id}"

    try:
        response = requests.get(url, timeout=10)
        response.raise_for_status()
        data = response.json()

        # Extract the ChEMBL ID from the response structure
        if isinstance(data, list) and len(data) > 0:
            return data[0].get("src_compound_id")
        return None
    except requests.exceptions.RequestException as e:
        print(f"UniChem API Error: {e}")
        return None
```
