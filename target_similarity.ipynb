```python
#read df

import pandas as pd


df = pd.read_csv("mapped_targets_with_fasta.csv")


df = df.dropna()
df["drug"] = df["drug"].str.strip()
df["uniprot_id"] = df["uniprot_id"].str.strip()

print(df.head(6))

# new grouyps with dug name plus its uniprot id(s)
drug_targets = df.groupby("drug")["uniprot_id"].apply(set).to_dict()

# Jaccard simialrity
def jaccard(set1, set2):
    intersection = set1 & set2
    union = set1 | set2
    
    if len(union) == 0:
        return 0
    
    return len(intersection) / len(union)

drugs = list(drug_targets.keys())

# imilarity matrix
similarity = {}

for drug1 in drugs:
    similarity[drug1] = {}
    
    for drug2 in drugs:
        score = jaccard(drug_targets[drug1], drug_targets[drug2])
        similarity[drug1][drug2] = score

# convert to df
similarity_df = pd.DataFrame(similarity)

# new cvs with similarity
similarity_df.to_csv("drug_jaccard_matrix.csv")
```
